```python
import pandas as pd

df = pd.read_csv('data/master_weekly.csv')

pd.set_option('display.max_columns', None)

df.tail(100)
df =df.drop(columns=['date'])

df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

#correlation matrix




fig = plt.figure(figsize=(12, 10))

sns.heatmap(df.corr(), annot=False, cmap='coolwarm', fmt='.2f', linewidths=0.5)
```

```python
import numpy as np

MAX_CORR = 0.95
MIN_CORR = 0.1



abs_corr = df.corr().abs()
mask = np.triu(np.ones(abs_corr.shape, dtype=bool), k=1)
corr_pairs = abs_corr.where(mask).stack().sort_values(ascending=False)

pairs = corr_pairs[corr_pairs.between(MIN_CORR, MAX_CORR, inclusive="neither")]
display(pairs)
```
