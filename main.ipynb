```python
import numpy as np
import pandas as pd
from sklearn.base import clone
from sklearn.compose import TransformedTargetRegressor
from sklearn.cross_decomposition import PLSRegression
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.linear_model import LinearRegression, ElasticNet
from sklearn.model_selection import GridSearchCV, PredefinedSplit
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score, mean_absolute_error


# ============================================================================
# CONFIG
# ============================================================================
MAX_FEAT_CORR   = 0.95
MIN_TARGET_CORR = 0.05
N_PRED, N_TEST, N_VAL = 26, 52, 52
REFIT_ON_TRAIN_VAL = True
DATE_COLS = ['month', 'quarter']

USE_HARMONICS        = True
USE_TARGET_CALENDAR  = True
N_HARM               = 3

horizons = {
    '2w':  {'y': 'y_vs_chg_2w',  'freqs': ['hebdo', 'journalier', 'calendrier']},
    '4w':  {'y': 'y_vs_chg_4w',  'freqs': ['hebdo', 'journalier', 'calendrier']},
    '8w':  {'y': 'y_vs_chg_8w',  'freqs': ['hebdo', 'journalier', 'calendrier', 'mensuel']},
    '13w': {'y': 'y_vs_chg_13w', 'freqs': ['hebdo', 'journalier', 'calendrier', 'mensuel']},
    '26w': {'y': 'y_vs_chg_26w', 'freqs': ['hebdo', 'journalier', 'calendrier', 'mensuel', 'trimestriel']},
}
TARGET_COLS = [cfg['y'] for cfg in horizons.values()]

ENET_GRID = {'regressor__m__alpha':    [0.01, 0.03, 0.1, 0.3, 1.0],
             'regressor__m__l1_ratio': [0.2, 0.5, 0.8, 1.0]}
PLS_K = [1, 2, 3, 4, 6]
ET_GRID = {'n_estimators': [300], 'max_depth': [3, 5, None],
           'min_samples_leaf': [5, 10, 20], 'max_features': [0.3, 0.6, 1.0]}
RF_GRID = {'n_estimators': [200], 'max_depth': [3, 5, 8, None],
           'min_samples_leaf': [1, 5, 10], 'max_features': [0.2, 0.5, 1.0]}
DT_GRID = {'max_depth': [2, 3, 5, 8, None], 'min_samples_leaf': [1, 5, 10, 20],
           'max_features': [0.5, 1.0]}


# ============================================================================
# 0. DATE HELPERS
# ============================================================================
def week_index(s):
    s = pd.to_datetime(pd.Series(s).astype(str).str[:10])
    return pd.DatetimeIndex(s.dt.to_period('W-SUN').dt.start_time, name='date')

def iso_to_monday(d):
    return pd.DatetimeIndex(
        pd.to_datetime(d['iso_year'].astype(str) + '-W' +
                       d['iso_week'].astype(int).astype(str).str.zfill(2) + '-1',
                       format='%G-W%V-%u'), name='date')


# ============================================================================
# 0b. DIRECTIONAL ACCURACY
# ============================================================================
def directional_accuracy(y_true, y_pred, tol=0.0):
    """Fraction of correct sign predictions. Weeks with |y_true| <= tol are
    ignored (no directional information). 0.5 = random, 1.0 = perfect."""
    y_true = np.asarray(y_true, float)
    y_pred = np.asarray(y_pred, float)
    mask = np.abs(y_true) > tol
    if mask.sum() == 0:
        return np.nan
    return float(np.mean(np.sign(y_true[mask]) == np.sign(y_pred[mask])))


# ============================================================================
# 1. ALIGN FEATURES AND LABELS BY DATE
# ============================================================================
df_label = pd.read_csv('data_seb/labels_valais_weekly.csv')
df_label.index = week_index(df_label['date'])
df_label = df_label.drop(columns='date')
df_label = df_label[~df_label.index.duplicated()]

df_dict = pd.read_csv('data/master_dictionary.csv')

feat = df.copy()
if isinstance(feat.index, pd.DatetimeIndex):
    feat.index = week_index(feat.index.to_series())
elif 'date' in feat.columns:
    feat.index = week_index(feat['date']); feat = feat.drop(columns='date')
else:
    feat.index = iso_to_monday(feat)
feat = feat[~feat.index.duplicated()].sort_index()
feat = feat.reindex(pd.date_range(feat.index.min(), feat.index.max(), freq='W-MON', name='date'))

CAL_COLS = [c for c in df_dict.loc[df_dict['native_frequency'] == 'calendrier', 'column']
            if c in feat.columns and c not in DATE_COLS and c not in ('iso_year', 'iso_week')]
if not CAL_COLS:
    CAL_COLS = [c for c in feat.columns if any(k in c for k in ('holiday', 'working_days', 'closure'))]
print('calendar columns:', CAL_COLS)

woy = feat.index.isocalendar().week.astype(float).values
HARM_COLS = []
for k in range(1, N_HARM + 1):
    feat[f'harm_sin{k}'] = np.sin(2 * np.pi * k * woy / 52.18)
    feat[f'harm_cos{k}'] = np.cos(2 * np.pi * k * woy / 52.18)
    HARM_COLS += [f'harm_sin{k}', f'harm_cos{k}']

TW_COLS = {}
for name in horizons:
    h = int(name[:-1])
    TW_COLS[name] = []
    for c in CAL_COLS:
        col = f'{c}__t{h}'
        feat[col] = feat[c].shift(-h)
        TW_COLS[name].append(col)

data = feat.join(df_label, how='inner')
print(f'aligned: {data.shape[0]} rows, {data.index.min().date()} -> {data.index.max().date()}')


# ============================================================================
# 2. SPLIT
# ============================================================================
n = len(data)
i_val, i_test, i_fut = n - (N_PRED + N_TEST + N_VAL), n - (N_PRED + N_TEST), n - N_PRED
train, val, test, future = data.iloc[:i_val], data.iloc[i_val:i_test], data.iloc[i_test:i_fut], data.iloc[i_fut:]
for nm, d in [('train', train), ('val', val), ('test', test), ('future', future)]:
    print(f'{nm:7s} {len(d):4d} rows  {d.index.min().date()} -> {d.index.max().date()}')

derived = set(HARM_COLS) | {c for v in TW_COLS.values() for c in v}
feature_cols = [c for c in feat.select_dtypes('number').columns
                if c not in TARGET_COLS and c not in DATE_COLS and c not in df_label.columns
                and c not in derived]


# ============================================================================
# 3. CORRELATION FILTER
# ============================================================================
corr_to_target = train[feature_cols + TARGET_COLS].corr().abs().loc[feature_cols, TARGET_COLS]
relevance = corr_to_target.mean(axis=1).sort_values(ascending=False)
feat_corr = train[feature_cols].corr().abs()

kept, dropped = [], {}
for col, rel in relevance.items():
    if np.isnan(rel) or rel < MIN_TARGET_CORR:
        dropped[col] = f'weak target corr ({rel:.2f})'
        continue
    if kept:
        worst = feat_corr.loc[col, kept].max()
        if worst > MAX_FEAT_CORR:
            dropped[col] = f'redundant with {feat_corr.loc[col, kept].idxmax()} ({worst:.2f})'
            continue
    kept.append(col)

print(f'features: {len(feature_cols)} -> kept {len(kept)}, dropped {len(dropped)}')
display(relevance.loc[kept].rename('mean |corr| to targets').to_frame())
display(pd.Series(dropped, name='reason').to_frame())

df_cleaned = data[kept + HARM_COLS + [c for v in TW_COLS.values() for c in v] + list(df_label.columns)]


# ============================================================================
# 4. MODELS
# ============================================================================
def cols_for(freqs):
    cols = df_dict.loc[df_dict['native_frequency'].isin(freqs), 'column']
    return [c for c in cols if c in kept]

def tune(estimator, grid, tr, va, cols_x, col_y):
    X = pd.concat([tr[cols_x], va[cols_x]])
    y = pd.concat([tr[col_y], va[col_y]])
    fold = np.r_[np.full(len(tr), -1), np.zeros(len(va), dtype=int)]
    gs = GridSearchCV(estimator, grid, scoring='r2', cv=PredefinedSplit(fold), n_jobs=-1, refit=False)
    gs.fit(X, y)
    return gs.best_params_, gs.best_score_

def make_specs(n_feat):
    return [
        ('seas', LinearRegression(), {}, 'harm'),
        ('lin',  LinearRegression(), {}, 'all'),
        ('enet', TransformedTargetRegressor(
                    regressor=Pipeline([('sc', StandardScaler()), ('m', ElasticNet(max_iter=20000))]),
                    transformer=StandardScaler()), ENET_GRID, 'all'),
        ('pls',  Pipeline([('sc', StandardScaler()), ('m', PLSRegression())]),
                 {'m__n_components': [k for k in PLS_K if k <= n_feat]}, 'all'),
        ('et',   ExtraTreesRegressor(random_state=42, n_jobs=-1),  ET_GRID, 'all'),
        ('rf',   RandomForestRegressor(random_state=42, n_jobs=-1), RF_GRID, 'all'),
        ('dt',   DecisionTreeRegressor(random_state=42),            DT_GRID, 'all'),
    ]

models, best_params, val_rows, test_rows, info = {}, {}, [], [], []

for name, cfg in horizons.items():
    col_y = cfg['y']
    cols_x = list(cols_for(cfg['freqs']))
    if USE_HARMONICS:
        cols_x += HARM_COLS
    if USE_TARGET_CALENDAR:
        cols_x += TW_COLS[name]
    all_cols = cols_x + [c for c in HARM_COLS if c not in cols_x]

    tr = df_cleaned.loc[train.index, all_cols + [col_y]].dropna()
    va = df_cleaned.loc[val.index,   all_cols + [col_y]].dropna()
    te = df_cleaned.loc[test.index,  all_cols + [col_y]].dropna()
    fit_df = pd.concat([tr, va]) if REFIT_ON_TRAIN_VAL else tr

    vrow, trow = {'horizon': name}, {'horizon': name}
    val_preds, test_preds = {}, {}

    for tag, est, grid, inputs in make_specs(len(cols_x)):
        cx = HARM_COLS if inputs == 'harm' else cols_x
        params, v = tune(est, grid, tr, va, cx, col_y)
        best_params[f'{name}_{tag}'] = params
        m = clone(est).set_params(**params).fit(fit_df[cx], fit_df[col_y])
        models[f'{name}_{tag}'] = m

        # Validation
        pred_va = np.ravel(m.predict(va[cx]))
        vrow[tag] = v
        vrow[f'{tag}_mae'] = mean_absolute_error(va[col_y], pred_va)
        vrow[f'{tag}_dir'] = directional_accuracy(va[col_y], pred_va)
        val_preds[tag] = pred_va

        # Test
        pred_te = np.ravel(m.predict(te[cx]))
        trow[tag] = r2_score(te[col_y], pred_te)
        trow[f'{tag}_mae'] = mean_absolute_error(te[col_y], pred_te)
        trow[f'{tag}_dir'] = directional_accuracy(te[col_y], pred_te)
        test_preds[tag] = pred_te

    # ---- RF + ET ensemble ----
    pred_va_mean = (val_preds['rf'] + val_preds['et']) / 2
    pred_te_mean = (test_preds['rf'] + test_preds['et']) / 2

    vrow['rf_et_mean'] = r2_score(va[col_y], pred_va_mean)
    vrow['rf_et_mean_dir'] = directional_accuracy(va[col_y], pred_va_mean)

    trow['rf_et_mean'] = r2_score(te[col_y], pred_te_mean)
    trow['rf_et_mean_dir'] = directional_accuracy(te[col_y], pred_te_mean)

    val_rows.append(vrow); test_rows.append(trow)
    info.append({'horizon': name, 'n_train': len(tr), 'n_val': len(va), 'n_test': len(te), 'n_cols': len(cols_x)})

display(pd.DataFrame(info).set_index('horizon'))
res_val  = pd.DataFrame(val_rows).set_index('horizon')
res_test = pd.DataFrame(test_rows).set_index('horizon')

print('R2 on VALIDATION year (best grid score; optimistic):')
display(res_val.drop(columns=[c for c in res_val.columns if c.endswith(('_mae', '_dir'))]).round(3))
print('R2 on TEST year (honest):')
display(res_test.drop(columns=[c for c in res_test.columns if c.endswith(('_mae', '_dir'))]).round(3))

print('MAE on VALIDATION year:')
display(res_val[[c for c in res_val.columns if c.endswith('_mae')]].round(3))
print('MAE on TEST year (honest):')
display(res_test[[c for c in res_test.columns if c.endswith('_mae')]].round(3))

print('Directional accuracy on VALIDATION year (0.5 = random):')
display(res_val[[c for c in res_val.columns if c.endswith('_dir')]].round(3))
print('Directional accuracy on TEST year (honest; 0.5 = random):')
display(res_test[[c for c in res_test.columns if c.endswith('_dir')]].round(3))

print('\nBest hyper-parameters:')
for k, v in best_params.items():
    print(f'  {k}: {v}')
```
