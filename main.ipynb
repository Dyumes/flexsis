```python
import pandas as pd

df = pd.read_csv('data/master_weekly.csv')

pd.set_option('display.max_columns', None)

df.tail(100)
df =df.drop(columns=['date'])

df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

#correlation matrix




fig = plt.figure(figsize=(12, 10))

sns.heatmap(df.corr(), annot=False, cmap='coolwarm', fmt='.2f', linewidths=0.5)
```

```python
MAX_CORR = 0.95
MIN_CORR = 0.1

abs_corr = df.corr().abs()
corr_pairs = abs_corr.unstack().sort_values(ascending=False).loc[lambda x: x < 1]
good_corr_pairs = corr_pairs.loc[lambda x: x > MAX_CORR].drop_duplicates()
good_corr_pairs.add(corr_pairs.loc[lambda x: x < MIN_CORR].drop_duplicates())

# Get columns to drop based on high correlation
columns_to_drop = set()
for col1, col2 in good_corr_pairs.index:
    if col1 not in columns_to_drop:
        columns_to_drop.add(col2)  # Drop the second column in the pair
    if col2 not in columns_to_drop:
        columns_to_drop.add(col1)  # Drop the first column in the pair


display(good_corr_pairs)
display(columns_to_drop)

# Drop the columns with high correlation
df_cleaned = df.drop(columns=columns_to_drop)

# df_cleaned = df.drop(columns=columns_to_drop)
display(df_cleaned)
```
